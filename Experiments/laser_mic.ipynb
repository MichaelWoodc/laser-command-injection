```python
import pyaudio
import numpy as np
import matplotlib.pyplot as plt

# --- Audio config ---
CHUNK = 1024
RATE = 44100
FORMAT = pyaudio.paInt16
CHANNELS = 1

# --- PyAudio setup ---
p = pyaudio.PyAudio()
stream = p.open(
    format=FORMAT,
    channels=CHANNELS,
    rate=RATE,
    input=True,
    frames_per_buffer=CHUNK
)

# --- Matplotlib interactive window ---
plt.ion()
fig, ax = plt.subplots()
x = np.arange(0, CHUNK)
line, = ax.plot(x, np.zeros(CHUNK), lw=2)

ax.set_ylim(-32768, 32767)
ax.set_xlim(0, CHUNK)
ax.set_title("Real‑Time Microphone Waveform")
ax.set_xlabel("Samples")
ax.set_ylabel("Amplitude")

fig.show()
fig.canvas.draw()

print("Listening... Close the pop‑out window to stop.")

# --- Live update loop ---
while plt.fignum_exists(fig.number):
    data = stream.read(CHUNK, exception_on_overflow=False)
    audio = np.frombuffer(data, dtype=np.int16)
    line.set_ydata(audio)
    fig.canvas.draw()
    fig.canvas.flush_events()

# --- Cleanup ---
stream.stop_stream()
stream.close()
p.terminate()

print("Stopped.")
```
