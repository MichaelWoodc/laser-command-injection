```python
%pip install opencv-python

import cv2

print("OpenCV version:", cv2.__version__)
print("Scanning camera indexes 0-9 with DirectShow and Media Foundation...")
print("A preview will open for each readable device.")
print("Press n for the next camera, q to stop.")

backends = [
    ("DirectShow", cv2.CAP_DSHOW),
    ("Media Foundation", cv2.CAP_MSMF),
]
readable_cameras = []
stop_scan = False

cv2.namedWindow("Camera scan", cv2.WINDOW_NORMAL)
cv2.resizeWindow("Camera scan", 960, 720)

try:
    for backend_name, backend in backends:
        for index in range(10):
            camera = cv2.VideoCapture(index, backend)
            if not camera.isOpened():
                camera.release()
                continue

            ok, frame = camera.read()
            if not ok or frame is None:
                camera.release()
                continue

            height, width = frame.shape[:2]
            readable_cameras.append((backend_name, index, width, height))
            print(f"Readable camera: {backend_name}, index {index}, {width}x{height}")

            while True:
                ok, frame = camera.read()
                if not ok or frame is None:
                    print(f"Frame read failed for {backend_name}, index {index}")
                    break

                preview = frame.copy()
                label = f"{backend_name} | camera index {index} | {width}x{height}"
                cv2.putText(
                    preview,
                    label,
                    (15, 35),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.8,
                    (0, 255, 0),
                    2,
                    cv2.LINE_AA,
                )
                cv2.imshow("Camera scan", preview)
                key = cv2.waitKey(1) & 0xFF
                if key == ord("n"):
                    break
                if key == ord("q"):
                    stop_scan = True
                    break

            camera.release()
            if stop_scan:
                break
        if stop_scan:
            break
finally:
    cv2.destroyAllWindows()

print("\nScan complete.")
if readable_cameras:
    print("Readable devices:")
    for camera_info in readable_cameras:
        print("  ", camera_info)
else:
    print("No readable cameras found through these Windows backends.")
```
